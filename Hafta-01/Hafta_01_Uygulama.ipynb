## 1. Kullanıcı Giriş Verilerini İnceleyelim

Bir bilgi sistemine yapılan kullanıcı girişlerini takip ettiğimizi düşünelim.

Bir kullanıcının hesabına kısa süre içerisinde çok sayıda başarısız giriş yapılması, olası bir **brute-force (kaba kuvvet) saldırısının** göstergelerinden biri olabilir.

İlk olarak tek bir kullanıcıya ait başarısız giriş sayısını inceleyelim.

Aşağıdaki kodda:

- `kullanici` → giriş yapan kullanıcıyı,
- `basarisiz_giris` → başarısız giriş denemesi sayısını

temsil etmektedir.

```python
kullanici = "user01"
basarisiz_giris = 3

print("Kullanıcı:", kullanici)
print("Başarısız giriş sayısı:", basarisiz_giris)
```

## 2. Kural Tabanlı Güvenlik Sistemi

Şimdi sistemimizin otomatik olarak bir güvenlik kararı vermesini sağlayalım.

Güvenlik kuralımız şu olsun:

> **Bir kullanıcı 10'dan fazla başarısız giriş yapmışsa şüpheli olarak işaretle.**

Bu kuralı Python ile uygulayabiliriz.

```python
basarisiz_giris = 9

if basarisiz_giris > 10:
    print("⚠️ UYARI: Şüpheli giriş tespit edildi!")
else:
    print("✅ Normal kullanıcı davranışı.")
```

### 🤔 Düşünelim

Aşağıdaki soruları cevaplayınız:

1. `basarisiz_giris = 10` olduğunda sistem neden uyarı vermedi?
2. `basarisiz_giris = 11` olduğunda neden uyarı verdi?
3. Sistemin hangi durumda "şüpheli" kararı vereceğini kim belirledi?
4. Bu sistem verilerden herhangi bir şey öğrendi mi?

### Sonuç

Bu sistem **makine öğrenmesi değildir**.

Çünkü sistem:

- geçmiş giriş verilerini inceleyerek bir örüntü öğrenmedi,
- kendi karar sınırını oluşturmadı,
- yeni verilerden öğrenmedi.

`10` sınırını **biz belirledik**.

Yani:

**İnsan → Kuralı belirledi → Bilgisayar kuralı uyguladı**

Bu nedenle oluşturduğumuz sistem bir **kural tabanlı (rule-based) sistemdir**.

## 5. Model Neye Bakacak?

Tablomuzdaki her satır geçmişte gerçekleşmiş bir kullanıcı girişini temsil etmektedir.

Modelin kullanabileceği bilgiler:

- `Basarisiz_Giris`
- `Gece_Girisi`
- `Yeni_Cihaz`

Bu bilgilere **özellik (feature)** denir.

Modelin öğrenmesini istediğimiz sonuç ise:

- `Durum`

sütunudur.

Bu sütuna **etiket (label)** denir.

Yani:

**Feature'lar → Modelin karar verirken kullandığı bilgiler**

**Label → Modelin tahmin etmeye çalıştığı sonuç**

### Veri Yapımız

| Feature | Feature | Feature | Label |
|---|---|---|---|
| Başarısız Giriş | Gece Girişi | Yeni Cihaz | Durum |
| 1 | Hayır | Hayır | Normal |
| 18 | Evet | Evet | Saldırı |
| 25 | Evet | Evet | Saldırı |

Modelin amacı, feature'lar ile label arasındaki ilişkileri öğrenmektir.

## 6. Önce Biz Tahmin Edelim

Henüz bir makine öğrenmesi modeli oluşturmayacağız.

Önce tablodaki örnekleri kendimiz inceleyelim.

Aşağıdaki yeni kullanıcı girişinin durumunu tahmin etmeye çalışın:

- Başarısız giriş sayısı: **20**
- Gece girişi: **Evet**
- Yeni cihaz: **Evet**

### Sizce sonuç nedir?

- Normal
- Saldırı

Kararınızı verirken geçmiş tablodaki hangi örneklerden yararlandınız?

### Neden "Saldırı" Dedik?

Geçmiş örneklere baktığımızda;

- yüksek başarısız giriş sayısının,
- gece yapılan girişlerin,
- yeni cihaz kullanımının

saldırı olarak etiketlenmiş örneklerde daha sık görüldüğünü fark etmiş olabiliriz.

Yani aslında biz de geçmiş verilerdeki **örüntüleri (patterns)** kullanarak yeni bir örnek hakkında tahmin yaptık.

Makine öğrenmesinin temel fikri de buna benzer:

> **Geçmiş verilerdeki ilişkileri öğren ve daha önce görülmemiş veri üzerinde tahmin yap.**

## 7. Kural Tabanlı Sistem ile Makine Öğrenmesi Arasındaki Fark

### Kural Tabanlı Yaklaşım

Kuralı insan belirler:

`Başarısız giriş > 10 VE gece girişi VE yeni cihaz → Saldırı`

Akış:

**İnsan → Kural → Bilgisayar → Karar**

---

### Makine Öğrenmesi Yaklaşımı

Sisteme geçmiş örnekler verilir:

**Geçmiş Veriler → Öğrenme Algoritması → Model → Yeni Veri → Tahmin**

Burada amaç, bütün karar kurallarını tek tek bizim yazmamız yerine modelin verilerdeki ilişkileri öğrenmesidir.

| Kural Tabanlı Sistem | Makine Öğrenmesi |
|---|---|
| Kuralları insan yazar | Model verilerden ilişkiler öğrenir |
| Önceden belirlenmiş koşullara dayanır | Geçmiş örneklerden yararlanır |
| Yeni durumlar için yeni kurallar gerekebilir | Yeni veriler üzerinde tahmin yapabilir |
| Karmaşık problemlerde çok sayıda kural gerekebilir | Çok sayıda özelliği birlikte değerlendirebilir |

## 8. Peki Bilgisayar Bu İlişkileri Nasıl Öğrenecek?

Şu ana kadar önemli bir soruyu açık bıraktık:

> Bilgisayar geçmiş verilere bakarak bu ilişkileri **nasıl** öğrenecek?

Bunun için **makine öğrenmesi algoritmaları** kullanılır.

Örneğin ilerleyen haftalarda;

- Logistic Regression
- Decision Tree
- Random Forest

gibi sınıflandırma yöntemlerini inceleyeceğiz.

Bu algoritmalar kullanılarak eğitim verilerinden bir **model** oluşturulur.

Genel süreç:

**1. Veri Toplama**  
↓  
**2. Feature ve Label Belirleme**  
↓  
**3. Model Seçme**  
↓  
**4. Modeli Eğitme**  
↓  
**5. Yeni Veri Verme**  
↓  
**6. Tahmin Alma**

Bu hafta algoritmaların nasıl çalıştığını öğrenmeyeceğiz. Öncelikle bu sürecin mantığını anlamamız yeterlidir.
