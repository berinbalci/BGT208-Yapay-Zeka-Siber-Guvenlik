## 9. Eksik Verilerin Kontrol Edilmesi

Gerçek veri setlerinde bazı kayıtların belirli özellikleri eksik olabilir.

Örneğin bir sütunda değer bulunmaması, o gözleme ait bilginin
kaydedilmediği veya elde edilemediği anlamına gelebilir.

Makine öğrenmesi modellerini geliştirmeden önce eksik değerlerin
kontrol edilmesi önemlidir.

Pandas'ta `isnull()` eksik değerleri tespit eder.
`sum()` ile birlikte kullanıldığında her sütundaki toplam eksik
değer sayısını görebiliriz.

```python
df.isnull().sum()
```

```python
eksik_degerler = df.isnull().sum()

eksik_degerler[eksik_degerler > 0]
```

### Sonucu Yorumlayalım

- Sonucun `0` olması ilgili sütunda eksik değer bulunmadığını gösterir.
- Sıfırdan büyük bir değer, o sütunda eksik kayıtlar bulunduğunu gösterir.
- Eksik değerlerin bulunması durumunda bu değerlerin nasıl ele alınacağına
  veri setine ve probleme göre karar verilmelidir.

Bu aşamada eksik değerleri silmeye veya doldurmaya çalışmıyoruz.
Amacımız yalnızca veri setinde eksik veri olup olmadığını tespit etmektir.

## 10. Tekrarlanan Kayıtların Kontrol Edilmesi

Bir veri setinde aynı kayıt birden fazla kez bulunabilir.
Bu tür kayıtlara **tekrarlanan kayıt (duplicate)** adı verilir.

Tekrarlanan kayıtlar veri setindeki bazı örneklerin gereğinden fazla
temsil edilmesine neden olabilir. Bu nedenle veri analizi sırasında
kontrol edilmeleri önemlidir.

Pandas'ta `duplicated()` fonksiyonu tekrarlanan kayıtları belirler.
`sum()` ile birlikte kullanıldığında toplam tekrarlanan kayıt sayısını verir.

```python
df.duplicated().sum()
```

```python
df[df.duplicated()].head()
```

### Sonucu Yorumlayalım

- Sonuç `0` ise veri setinde tamamen aynı olan tekrarlanmış kayıt bulunmamaktadır.
- Sonuç `0`'dan büyükse veri setinde tekrarlanan kayıtlar bulunmaktadır.

Bu aşamada tekrarlanan kayıtları silmiyoruz.
Amacımız veri setinin kalitesini incelemek ve olası problemleri tespit etmektir.

```python
df["label"].value_counts()
```

```python
df["label"].value_counts(normalize=True) * 100
```

```python
print("Sınıf sayıları:")
print(df["label"].value_counts())

print("\nSınıf yüzdeleri:")
print(df["label"].value_counts(normalize=True) * 100)
```

```python
### Sonucu Yorumlayalım

`value_counts()` her sınıfa ait kayıt sayısını gösterir.

`value_counts(normalize=True)` ise sınıfların veri setindeki oranlarını
hesaplar.

Bir sınıfın diğer sınıfa göre çok daha fazla örnek içermesi
**sınıf dengesizliği (class imbalance)** olarak adlandırılır.

Sınıf dağılımını incelemek önemlidir çünkü sınıfların dağılımı,
ilerleyen haftalarda geliştireceğimiz makine öğrenmesi modellerinin
eğitilmesini ve değerlendirilmesini etkileyebilir.
```

## 13. Veri Görselleştirme

Verileri yalnızca tablo ve sayılarla incelemek yerine grafiklerle
göstermek, verinin yapısını daha kolay anlamamızı sağlar.

Python'da temel grafikler oluşturmak için **Matplotlib** kütüphanesi
kullanılabilir.

Bu örnekte veri setindeki `label` sınıflarının dağılımını
sütun grafiği ile göstereceğiz.

```python
import matplotlib.pyplot as plt

df["label"].value_counts().plot(kind="bar")

plt.xlabel("Sınıf")
plt.ylabel("Kayıt Sayısı")
plt.title("Phishing Veri Seti Sınıf Dağılımı")

plt.show()
```

### Grafiği Yorumlayalım

Grafikte her sütun bir sınıftaki kayıt sayısını göstermektedir.

Bu grafik sayesinde sınıflardan birinin diğerine göre daha fazla
örneğe sahip olup olmadığını görsel olarak değerlendirebiliriz.

### Düşünelim

- Veri setinde kaç sınıf bulunmaktadır?
- Hangi sınıfta daha fazla kayıt vardır?
- Sınıflar birbirine yakın sayıda örnek içeriyor mu?
- Sınıflar arasında büyük bir fark olsaydı bu durum ileride
  geliştireceğimiz makine öğrenmesi modelini nasıl etkileyebilirdi?

```python
siniflar = df["label"].map({
    0: "Phishing",
    1: "Legitimate"
})

siniflar.value_counts().plot(kind="bar")

plt.xlabel("Web Sitesi Sınıfı")
plt.ylabel("Kayıt Sayısı")
plt.title("Phishing ve Legitimate Web Sitelerinin Dağılımı")

plt.show()
```
