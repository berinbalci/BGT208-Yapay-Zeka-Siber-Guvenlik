## Siber Güvenlik Veri Seti Nasıl Aranır?

Bir makine öğrenmesi projesinde ilk adımlardan biri, probleme uygun
bir veri seti bulmaktır.

Veri seti ararken yalnızca **"cybersecurity dataset"** gibi genel ifadeler
kullanmak yerine, çözmek istediğimiz güvenlik problemini belirten
anahtar kelimeler kullanmak daha doğru sonuçlara ulaşmamızı sağlar.

### Örnek Arama Kelimeleri

| Problem | Arama örneği |
|---|---|
| Phishing tespiti | `phishing dataset` |
| Zararlı URL tespiti | `malicious URL dataset` |
| Ağ saldırısı tespiti | `network intrusion detection dataset` |
| Anomali tespiti | `network anomaly detection dataset` |
| Malware tespiti | `malware detection dataset` |
| Spam tespiti | `spam detection dataset` |
| DDoS saldırısı | `DDoS attack dataset` |
| Brute-force saldırısı | `brute force attack dataset` |

Arama yaparken İngilizce anahtar kelimeler kullanmak genellikle
daha fazla akademik ve açık veri kaynağına ulaşmayı sağlar.

## 3. UCI Machine Learning Repository'den Doğrudan Veri Alma

Bir veri setini önce bilgisayara indirip daha sonra Google Colab'a
yüklemek her zaman gerekli değildir.

Eğer veri setinin internet üzerinde doğrudan erişilebilen bir dosya adresi
varsa, bu dosya Python kullanılarak doğrudan çalışma ortamına alınabilir.

Bu uygulamada **PhiUSIIL Phishing URL Dataset** kullanılacaktır.

UCI Machine Learning Repository üzerinde bulunan veri seti,
phishing ve güvenilir web sitelerine ait URL özelliklerini içermektedir.

Bu yöntemde veri seti bilgisayara manuel olarak indirilmeyecek;
internet üzerindeki veri kaynağından doğrudan Google Colab ortamına
aktarılacaktır.

```python
import urllib.request

url = "https://archive.ics.uci.edu/static/public/967/phiusiil+phishing+url+dataset.zip"

urllib.request.urlretrieve(url, "phiusiil_dataset.zip")

print("Veri seti indirildi.")
```
